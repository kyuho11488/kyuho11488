```python
# 문제 1-1

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# 데이터 불러오기
data = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/data0405.csv', usecols=['x','y'])
X = data[['x']].values
y = data['y'].values

# 1. 최소제곱법 (Linear)

def least_squares_poly(degree):
    poly = PolynomialFeatures(degree=degree)
    X_poly = poly.fit_transform(X)
    model = LinearRegression()
    model.fit(X_poly, y)
    return model, poly

def plot_model(X, y, model, poly, title):
    x_range = np.linspace(X.min(), X.max(), 500).reshape(-1, 1)
    X_range_poly = poly.transform(x_range)
    y_pred = model.predict(X_range_poly)
    plt.scatter(X, y, color='blue', label='Data')
    plt.plot(x_range, y_pred, color='red', label='Model')
    plt.title(title)
    plt.xlabel('x')
    plt.ylabel('y')
    plt.legend()
    plt.grid(True)
    plt.show()

# 1-1 Linear
model_lin, poly_lin = least_squares_poly(1)
print("Least Squares - Linear:", model_lin.coef_, model_lin.intercept_)
plot_model(X, y, model_lin, poly_lin, "Least Squares - Linear Model")
```

```python
# 문제 2-1

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# 데이터 불러오기
data = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/data0405.csv', usecols=['x','y'])
X = data[['x']].values
y = data['y'].values

# 2. 경사하강법 (Linear)

# 선형 회귀 - 경사하강법
def gradient_descent_linear(X, y, lr=0.001, epochs=500): # lr = learning rate, epochs(반복횟수)
    n = len(y)
    w = 0.0
    b = 0.0
    for _ in range(epochs):
        y_pred = w * X + b
        error = y_pred - y
        dw = (2/n) * np.sum(error * X)
        db = (2/n) * np.sum(error)
        w -= lr * dw
        b -= lr * db
    return w, b

# 다항 회귀 - 경사하강법
def gradient_descent_poly(X, y, degree, lr=0.001, epochs=500): # lr = learning rate, epochs(반복횟수)
    poly = PolynomialFeatures(degree=degree)
    X_poly = poly.fit_transform(X)
    n_samples, n_features = X_poly.shape
    weights = np.zeros(n_features)

    for _ in range(epochs):
        y_pred = X_poly @ weights
        error = y_pred - y
        gradient = (2 / n_samples) * X_poly.T @ error
        weights -= lr * gradient
    return weights, poly

# 2-1 Linear
w_gd, b_gd = gradient_descent_linear(X.flatten(), y)
print("Gradient Descent - Linear:", w_gd, b_gd)
plt.scatter(X, y, color='blue')
plt.plot(X, w_gd * X + b_gd, color='orange')
plt.title('Gradient Descent - Linear')
plt.grid(True)
plt.show()
```

```python
# 문제 2-2

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# 데이터 불러오기
data = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/data0405.csv', usecols=['x','y'])
X = data[['x']].values
y = data['y'].values

# 2. 경사하강법 (2차, 4차)

# 선형 회귀 - 경사하강법
def gradient_descent_linear(X, y, lr=0.001, epochs=500): # lr = learning rate, epochs(반복횟수)
    n = len(y)
    w = 0.0
    b = 0.0
    for _ in range(epochs):
        y_pred = w * X + b
        error = y_pred - y
        dw = (2/n) * np.sum(error * X)
        db = (2/n) * np.sum(error)
        w -= lr * dw
        b -= lr * db
    return w, b

# 다항 회귀 - 경사하강법
def gradient_descent_poly(X, y, degree, lr=0.001, epochs=500): # lr = learning rate, epochs(반복횟수)
    poly = PolynomialFeatures(degree=degree)
    X_poly = poly.fit_transform(X)
    n_samples, n_features = X_poly.shape
    weights = np.zeros(n_features)

    for _ in range(epochs):
        y_pred = X_poly @ weights
        error = y_pred - y
        gradient = (2 / n_samples) * X_poly.T @ error
        weights -= lr * gradient
    return weights, poly

# 2-2 2차
weights_2, poly2 = gradient_descent_poly(X, y, degree=2)
print("Gradient Descent - 2nd Degree:", weights_2)
x_range = np.linspace(X.min(), X.max(), 500).reshape(-1, 1)
y_pred_2 = poly2.transform(x_range) @ weights_2
plt.scatter(X, y, color='blue')
plt.plot(x_range, y_pred_2, color='green')
plt.title("Gradient Descent - 2nd Degree")
plt.grid(True)
plt.show()

# 2-2 4차
weights_4, poly4 = gradient_descent_poly(X, y, degree=4)
print("Gradient Descent - 4th Degree:", weights_4)
y_pred_4 = poly4.transform(x_range) @ weights_4
plt.scatter(X, y, color='blue')
plt.plot(x_range, y_pred_4, color='purple')
plt.title("Gradient Descent - 4th Degree")
plt.grid(True)
plt.show()
```
